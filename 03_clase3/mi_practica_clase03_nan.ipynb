Ejercicio 1.1: Carga del Dataset e Inspección Estructural (read_csv, shape e info)

```python
import pandas as pd
import numpy as np

df = pd.read_csv("C:\\Users\\adria\\Desktop\\00_fapd_total\\00_fapd\\00_DATA_Clase03_Archivos_Actividades\\DATA_Clientes_NaN.csv")

df1 = df.copy()

print(df1.head(8))
print(df1.info())
```

▸ Ejercicio 1.2: Auditoría Cuantitativa de Valores Faltantes (isna y porcentajes)

```python
print("Máscara booleana de ausentes (primeras 10 filas; True = faltante, False = presente):")
display(df.isna().head(10))

print("Valores NaN por variable:")
display(df.isna().sum().rename("cantidad_NaN"))

print("Total global de celdas ausentes:")
print(df.isna().sum().sum())

print("Porcentaje de incompletitud por columna (%):")
display((df.isna().mean() * 100).round(2).rename("incompletitud_%"))
```

Ejercicio 1.3: Demostración Empírica del Riesgo Matemático

```python
df["ahorro_bruto"] = df["ingresos_mensuales"] - df["gasto_mensual"]

filas_sin_ahorro = df[df["ahorro_bruto"].isna()]
print("Filas cuyo ahorro bruto resulta NaN:")
display(filas_sin_ahorro)

print("NaN en ingresos_mensuales entre esas filas:", filas_sin_ahorro["ingresos_mensuales"].isna().sum())
print("NaN en gasto_mensual entre esas filas:", filas_sin_ahorro["gasto_mensual"].isna().sum())

df.drop("ahorro_bruto", axis=1, inplace=True)
```

▸ Ejercicio 2.1: Impacto de la Eliminación Global (dropna por filas y por columnas)

```python
df_sin_filas = df.dropna()
filas_supervivientes = len(df_sin_filas)
porcentaje_perdido = (24 - filas_supervivientes) / 24 * 100
print("Filas que sobreviven:", filas_supervivientes)
print(f"Porcentaje del volumen original perdido: {porcentaje_perdido:.2f}%")

df_sin_cols = df.dropna(axis=1)
print("Columnas que sobreviven:", df_sin_cols.columns.tolist())
display(df_sin_cols)

print("La eliminación indiscriminada por filas descarta clientes con cualquier dato ausente; por columnas, elimina variables potencialmente útiles. En este dataset, ninguna de las dos estrategias conserva toda la información relevante.")
```

Ejercicio 2.2: Eliminación Selectiva y Quirúrgica (drop)

```python
df_limpio = df.drop(["codigo_referido_antiguo"], axis=1)
print("Dimensiones de df_limpio:", df_limpio.shape)

df_prueba = df_limpio.copy()
print("Dimensiones iniciales de df_prueba:", df_prueba.shape)

df_prueba = df_prueba.drop([3, 7])
print("Dimensiones de df_prueba tras eliminar los índices 3 y 7:", df_prueba.shape)
```

 Ejercicio 3.1: Imputación por Valor Fijo (fillna(0))


```python
df_imputado = df_limpio.copy()
df_imputado["bono_fidelidad"] = df_imputado["bono_fidelidad"].fillna(0)

print("Valores ausentes restantes en bono_fidelidad:", df_imputado["bono_fidelidad"].isna().sum())
```

 Ejercicio 3.2: Imputación Secuencial en Fechas (Forward Fill vs. Back Fill)

```python
indices = [2, 7, 14]
fechas = df_imputado["fecha_registro"]
comparacion_fechas = pd.DataFrame({
    "original": fechas.loc[indices],
    "ffill": fechas.ffill().loc[indices],
    "bfill": fechas.bfill().loc[indices],
})
display(comparacion_fechas)

df_imputado["fecha_registro"] = df_imputado["fecha_registro"].ffill()
print("Valores ausentes restantes en fecha_registro:", df_imputado["fecha_registro"].isna().sum())
```

▸ Ejercicio 3.3: Imputación por Medidas de Centralidad (Media, Mediana y Moda)


```python
media_edad = df_imputado["edad"].mean()
print("Media de edad:", media_edad.round(2))
df_imputado["edad"] = df_imputado["edad"].fillna(df_imputado["edad"].mean())

print("Ingresos - media:", df_imputado["ingresos_mensuales"].mean().round(2))
print("Ingresos - mediana:", df_imputado["ingresos_mensuales"].median().round(2))
df_imputado["ingresos_mensuales"] = df_imputado["ingresos_mensuales"].fillna(df_imputado["ingresos_mensuales"].median())
df_imputado["gasto_mensual"] = df_imputado["gasto_mensual"].fillna(df_imputado["gasto_mensual"].median())

moda_ciudad = df_imputado["ciudad"].mode()[0]
print("Moda de ciudad:", moda_ciudad)
df_imputado["ciudad"] = df_imputado["ciudad"].fillna(moda_ciudad)

print("Valores ausentes por variable:")
display(df_imputado.isna().sum())
print("Número de filas conservadas:", len(df_imputado))
df_imputado.info()
```
