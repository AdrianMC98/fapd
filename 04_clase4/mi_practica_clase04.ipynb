# 01 EXPLORACIÓN ESTRUCTURAL Y AUDITORÍA DEL DATASET (nycflights13)
Ejercicio 1.1: Carga del Dataset e Inspección Inicial (dir, head, info y dtypes)


```python
import nycflights13
import pandas as pd
from nycflights13 import flights

print(dir(nycflights13))

df = flights.copy()

display(df.head())
print("Dimensiones:", df.shape)
df.info()
print(df.dtypes)
```

▸ Ejercicio 1.2: Auditoría Cuantitativa de Valores Faltantes (isna)


```python
total_nan = df.isna().sum().sum()
porcentaje_incompletitud = (df.isna().mean() * 100).round(2)

print("Total de valores NaN:", total_nan)
display(porcentaje_incompletitud.rename("Porcentaje de NaN (%)"))
```

02 TRANSFORMACIÓN, FILTRADO Y AGREGACIÓN DE DATOS CON PANDAS


```python
# 1. Vuelos del 1 de enero y vuelos de UA con retraso de salida >= 400 minutos
vuelos_1_enero = df[(df["month"] == 1) & (df["day"] == 1)]
vuelos_ua_retraso = df[(df["carrier"] == "UA") & (df["dep_delay"] >= 400)]

print("Vuelos del 1 de enero:", len(vuelos_1_enero))
display(vuelos_1_enero.head())

print("Vuelos de UA con retraso >= 400 minutos:", len(vuelos_ua_retraso))
display(vuelos_ua_retraso)

# 2. Vuelos de los dos primeros días de cada mes: ambas formas de filtrado
vuelos_dias_or = df[(df["day"] == 1) | (df["day"] == 2)]
vuelos_dias_between = df[df["day"].between(1, 2)]

print("¿Los resultados son iguales?:", vuelos_dias_or.equals(vuelos_dias_between))
display(vuelos_dias_between.head())

# 3. Vuelos con destino a Houston (IAH o HOU)
vuelos_houston = df[df["dest"].isin(["IAH", "HOU"])]
display(vuelos_houston.head())

# 4. Tres mayores y tres menores retrasos de salida
tres_mayores_retrasos = df.nlargest(3, "dep_delay")
tres_menores_retrasos = df.nsmallest(3, "dep_delay")

display(tres_mayores_retrasos)
display(tres_menores_retrasos)

# 5. Reutiliza el filtro de vuelos de UA con retraso >= 400 minutos
display(vuelos_ua_retraso)
```

▸ Ejercicio 2.2: Ordenación, Selección por Patrones y Renombrado (sort_values, .loc, rename)


```python
# 1. Ordenar por retraso de salida, de menor a mayor
vuelos_menor_retraso = df.sort_values(by="dep_delay", ascending=True)
display(vuelos_menor_retraso.head())

# 2. Ordenar por retraso de salida descendente y, en caso de empate,
#    por retraso de llegada ascendente
vuelos_ordenados = df.sort_values(
    by=["dep_delay", "arr_delay"],
    ascending=[False, True]
)
display(vuelos_ordenados.head())

# 3. Seleccionar columnas por nombre, rango y patrones
columnas_nombre = df[["year", "month", "day"]]
columnas_rango = df.loc[:, "year":"day"]
columnas_empiezan_a = df.loc[:, df.columns.str.startswith("a")]
columnas_terminan_y = df.loc[:, df.columns.str.endswith("y")]
columnas_contienen_ho = df.loc[:, df.columns.str.contains("ho")]

display(columnas_nombre.head())
display(columnas_rango.head())
display(columnas_empiezan_a.head())
display(columnas_terminan_y.head())
display(columnas_contienen_ho.head())

# 4. Renombrar dep_time en una copia, sin modificar df
df_ren = df.copy()
df_ren.rename(columns={"dep_time": "hora_de_salida"}, inplace=True)
display(df_ren[["hora_de_salida"]].head())
```

▸ Ejercicio 2.3: Creación de Nuevas Variables (Feature Creation)


```python
# Velocidad media en metros por segundo
df["velocidad_media"] = (df["distance"] * 1609.34) / (df["air_time"] * 60)

# Minutos recuperados durante el trayecto
df["ganancia"] = df["dep_delay"] - df["arr_delay"]

display(df[["distance", "air_time", "velocidad_media", "dep_delay", "arr_delay", "ganancia"]].head())
```

▸ Ejercicio 2.4: Agrupación y Resúmenes Estadísticos (groupby y agg)


```python
# 1. Media y desviación estándar de dep_delay por fecha
resumen_diario = (
    df.groupby(["year", "month", "day"])
      .agg(
          media_dep_delay=("dep_delay", "mean"),
          desviacion_std_dep_delay=("dep_delay", "std")
      )
      .reset_index()
)
display(resumen_diario.head())

# 2. Moda de dep_delay por aerolínea; si hay empate, se toma la primera
moda_por_aerolinea = df.groupby("carrier").agg(
    moda_dep_delay=(
        "dep_delay",
        lambda x: x.mode().iloc[0] if not x.mode().empty else None
    )
)
display(moda_por_aerolinea)

# 3. Comparar mínimo y máximo con first y last para dep_time por fecha
comparacion_horarios = df.groupby(["year", "month", "day"]).agg(
    dep_time_min=("dep_time", "min"),
    dep_time_max=("dep_time", "max"),
    dep_time_first=("dep_time", "first"),
    dep_time_last=("dep_time", "last")
).reset_index()
display(comparacion_horarios.head())

# 4. Destinos con mayor número de aerolíneas distintas
aerolineas_por_destino = (
    df.groupby("dest")
      .agg(aerolineas=("carrier", "nunique"))
      .reset_index()
      .sort_values(by="aerolineas", ascending=False)
)
display(aerolineas_por_destino.head())
```

# 03 CODIFICACIÓN DE VARIABLES CATEGÓRICAS Y ESCALADO NUMÉRICO
▸ Ejercicio 3.1: Codificación de Variables Categóricas (cat.codes, get_dummies y map)


```python
# 1. Comprobar las aerolíneas únicas y codificarlas numéricamente
print("Aerolíneas únicas:", df["carrier"].unique())
print("Número de aerolíneas:", df["carrier"].nunique())

df["carrier_code"] = df["carrier"].astype("category").cat.codes
print("Tipo de carrier_code:", df["carrier_code"].dtypes)
display(df["carrier_code"].head())

# 2. One-Hot Encoding: crea una columna binaria por aerolínea
flights_onehot = pd.get_dummies(df, columns=["carrier"], dtype=int)
columnas_onehot = [col for col in flights_onehot.columns if col.startswith("carrier_")]

print("Columnas binarias generadas:", len(columnas_onehot))
display(flights_onehot.head())

# 3. Aplicar el diccionario personalizado
mapa_carrier = {"UA": 0, "AA": 1, "B6": 2, "DL": 3}
carrier_mapeado = df["carrier"].map(mapa_carrier)

display(carrier_mapeado.head())
print("Valores NaN por aerolíneas omitidas:", carrier_mapeado.isna().sum())
```

▸ Ejercicio 3.2: Estandarización (StandardScaler) y Normalización (MinMaxScaler)


```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

cols = ["dep_delay", "arr_delay", "air_time", "distance"]
df_num = df[cols].dropna().copy()

# Estandarización Z-score
scaler_std = StandardScaler()
df_std = pd.DataFrame(
    scaler_std.fit_transform(df_num),
    columns=[c + "_std" for c in cols],
    index=df_num.index
)
display(df_std.describe().round(2))

# Normalización Min-Max
scaler_minmax = MinMaxScaler()
df_norm = pd.DataFrame(
    scaler_minmax.fit_transform(df_num),
    columns=[c + "_norm" for c in cols],
    index=df_num.index
)
print("Mínimos:", df_norm.min().round(2).to_dict())
print("Máximos:", df_norm.max().round(2).to_dict())
```

# 04 CONVERSIÓN DE TIPOS DE DATOS (dtypes) Y EXPORTACIÓN DE RESULTADOS
▸ Ejercicio 4.1: Conversión Numérica y Parseo Temporal (to_numeric y to_datetime)


```python
# Conversión numérica robusta: los valores inválidos se convierten en NaN
df["dep_delay"] = pd.to_numeric(df["dep_delay"], errors="coerce")
print("Tipo de dep_delay:", df["dep_delay"].dtype)

# Comprobación con un texto inválido
prueba = pd.to_numeric(pd.Series(["12", "texto inválido"]), errors="coerce")
display(prueba)

# Convertir las horas HHMM a objetos datetime.time
cols_datetime = ["dep_time", "sched_dep_time", "arr_time", "sched_arr_time"]

for col in cols_datetime:
    df[col] = pd.to_datetime(
        df[col].astype("Int64").astype(str).str.zfill(4),
        format="%H%M",
        errors="coerce"
    ).dt.time

# Combinar fecha y hora programada en una columna datetime
df["datetime"] = pd.to_datetime(df[["year", "month", "day", "hour", "minute"]])

display(df[cols_datetime + ["datetime"]].head())
print(df[cols_datetime + ["datetime"]].dtypes)
```

▸ Ejercicio 4.2: Optimización de Memoria con Tipo Categórico (category)

```python
cols_cat = ["carrier", "tailnum", "origin", "dest", "flight"]

print("Antes de convertir:")
df.info(memory_usage="deep")
memoria_antes = df.memory_usage(deep=True).sum()

for col in cols_cat:
    df[col] = df[col].astype("category")

print("\nDespués de convertir:")
df.info(memory_usage="deep")
memoria_despues = df.memory_usage(deep=True).sum()

print(f"\nMemoria antes:   {memoria_antes / 1024**2:.2f} MiB")
print(f"Memoria después: {memoria_despues / 1024**2:.2f} MiB")
print(f"Diferencia:      {(memoria_antes - memoria_despues) / 1024**2:.2f} MiB")
```
